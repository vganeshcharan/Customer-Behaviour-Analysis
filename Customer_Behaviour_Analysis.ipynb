**Data Cleaning**
'replacing null values'

```python
df.isnull().sum()
```

#***replacing Null values to median in review Rating cloumn***

```python
df['Review Rating'] = df.groupby('Category')['Review Rating'].transform(lambda x: x.fillna(x.median()))
```

```python
df.isnull().sum()
```

```python
df.columns = df.columns.str.lower()
df.columns = df.columns.str.replace(" ","_")
df = df.rename(columns={'purchase_amount_(usd)':'purchase_amount'})
```

```python
df.columns 
```

```python
#creating a age_group Column
labels = ['Young adult','Adult','Middle aged','Senior']
df['age_group'] = pd.qcut(df['age'],q=4,labels = labels)
```

```python
df[['age','age_group']].tail(10)
```

```python
#creating a column purchase_frequency_days
frequency_mapping = {
    'Fortnightly' : 14,
    'Weekly' : 7,
    'Monthly' : 30,
    'Quarterly' : 90,
    'Bi-Weekly' : 14,
    'Annually' : 365,
    'Every 3 Months' : 90
}
df['purchase_frequency_days'] = df['frequency_of_purchases'].map(frequency_mapping)
```

```python
df[['frequency_of_purchases' , 'purchase_frequency_days']].tail(10)
```

```python
(df['discount_applied'] == df['promo_code_used']).all()
```

**droping unnessary columns**

```python
df = df.drop('promo_code_used', axis=1)
```
